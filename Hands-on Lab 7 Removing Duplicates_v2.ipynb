```python
## Write your code here
# Create an empty list to store the nomarlized annual values
annual_comp = []

for freq, total in zip(df['CompFreq'], df['CompTotal']):
    if freq == 'Yearly':
        annual_comp.append(total)
    elif freq == 'Monthly':
        annual_comp.append(total * 12)
    elif freq == 'Weekly':
        annual_comp.append(total * 52)
    else:
        annual_comp.append(None)

df['NormalizedAnnualCompensation'] = annual_comp

print(f"Median Normalized Annual Compensation: {df['NormalizedAnnualCompensation'].median()}")
```
